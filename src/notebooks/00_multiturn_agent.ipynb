# 00 — Multi-turn code-action agent

Demonstrates explicit structured **Thought**, Python **Act**, `send_message`, persistent variables, and the next user message becoming the next Observation.

```python
# Install from the project root if needed:
# %pip install -e '.[gemini,notebook]'
```

```python
import os
from smolagents import OpenAIModel
from core import Agent, tool

model = OpenAIModel(
    model_id='gemini-2.5-flash',
    api_base='https://generativelanguage.googleapis.com/v1beta/openai/',
    api_key=os.environ['GOOGLE_API_KEY'],
    temperature=0,
)
```

```python
@tool
def echo(value: str) -> str:
    """Return the supplied text.

    Args:
        value: Text to echo.
    """
    return value

agent = Agent(model=model, tools=[echo], verbosity_level=1)
```

```python
first = agent.send(
    'Echo the text in text_0 and send it back.',
    objects={'text_0': 'hello'},
)
print(first.message.text)
```

```python
second = agent.send(
    'Now append an exclamation mark to the previous text and send it back.'
)
print(second.message.text)
```

```python
print(agent.trajectory_text)
```
